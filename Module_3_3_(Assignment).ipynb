## Question 1: Enumerate with List


```python
fruits = ["apple", "banana", "cherry", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

## Question 2: Zip with Two Lists

```python
students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]

for student, score in zip(students, scores):
    print(f"{student} scored {score}")
```

## Question 3: Tuple and Loop

```python
subjects = ("Math", "Physics", "Chemistry", "Biology", "English")

for subject in subjects:
    print(subject)
```

## Question 4: Set Operations

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

# Common elements
common_elements = set1 & set2
print("Common elements:", common_elements)

# Unique elements of set1
unique_set1 = set1 - set2
print("Unique elements of set1:", unique_set1)
```

## Question 5: Dictionary with Enumerate

```python
marks = {
    "Math": 80,
    "Physics": 75,
    "Chemistry": 85
}

for index, (subject, mark) in enumerate(marks.items(), start=1):
    print(f"{index} {subject}: {mark}")
```
